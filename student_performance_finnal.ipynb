```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()
y_train_encoded = encoder.fit_transform(y_train)
y_test_encoded = encoder.transform(y_test)
print(dict(zip(encoder.classes_, range(len(encoder.classes_)))))
```

```python
num_cols = X_train.select_dtypes(include='number').columns.tolist()
cat_cols = ['Revision Frequency', 'Phone During Study']

revision = ['Never', 'Only Before Exam', 'Monthly', 'Weekly', 'Daily']
phone = ['Actively Use', 'Silent', 'Keep Away']
```

### Preprocessing (imputation + encoding)

```python
from sklearn.compose import ColumnTransformer
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer, SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OrdinalEncoder

categorical = Pipeline([
    ('impute', SimpleImputer(strategy='most_frequent')),
    ('encode', OrdinalEncoder(categories=[revision, phone]))
])

def preprocessor():
    transformer = ColumnTransformer([
        ('num', IterativeImputer(random_state=42, max_iter=10), num_cols),
        ('cat', categorical, cat_cols)
    ], verbose_feature_names_out=False)
    transformer.set_output(transform='pandas')
    return transformer
```

### Clipping values after imputation

```python
from sklearn.preprocessing import FunctionTransformer

percentages = ['Attendance (%)', 'Assignment Submission (%)', 'Internal Marks (%)', 'Lab Score (%)']
discrete = ['No. of Backlogs', 'Active Backlogs', 'Online Certifications']
positive = [column for column in num_cols if column not in percentages + ['Sleep (hrs/day)']]

def limit(data):
    data = data.copy()
    data[percentages] = data[percentages].clip(0, 100)
    data['Sleep (hrs/day)'] = data['Sleep (hrs/day)'].clip(0, 12)
    data['Current CGPA (0 10)'] = data['Current CGPA (0 10)'].clip(upper=10)
    data[discrete] = data[discrete].round()
    data[positive] = data[positive].clip(lower=0)
    return data
```

```python
def build():
    return Pipeline([
        ('preprocess', preprocessor()),
        ('clip', FunctionTransformer(limit))
    ])

# quick check
check = build()
transformed = check.fit_transform(X_train)
print(transformed.isnull().sum().sum())
transformed.head()
```

```python
from sklearn.utils.class_weight import compute_sample_weight

weights = compute_sample_weight(class_weight='balanced', y=y_train_encoded)
```

### Building a pipeline for each model

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
from xgboost import XGBClassifier

lr = Pipeline([
    ('preprocessing', build()),
    ('scaler', StandardScaler()),
    ('model', LogisticRegression(max_iter=2000, class_weight='balanced', random_state=42))
])

rf = Pipeline([
    ('preprocessing', build()),
    ('model', RandomForestClassifier(n_estimators=300, class_weight='balanced', random_state=42, n_jobs=-1))
])

xgb = Pipeline([
    ('preprocessing', build()),
    ('model', XGBClassifier(n_estimators=300, learning_rate=0.05, max_depth=6,
                            objective='multi:softprob', num_class=len(encoder.classes_),
                            eval_metric='mlogloss', random_state=42))
])
```

### Cross-validation

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score

kfold = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

scores = {}
scores['Logistic Regression'] = cross_val_score(lr, X_train, y_train_encoded, cv=kfold, scoring='f1_macro')
scores['Random Forest'] = cross_val_score(rf, X_train, y_train_encoded, cv=kfold, scoring='f1_macro')
```

```python
# XGBoost doesn't have class_weight, so balance each fold by hand
from sklearn.base import clone
from sklearn.metrics import f1_score

folds = []
for train, valid in kfold.split(X_train, y_train_encoded):
    Xtrain, Xvalid = X_train.iloc[train], X_train.iloc[valid]
    ytrain, yvalid = y_train_encoded[train], y_train_encoded[valid]
    weight = compute_sample_weight(class_weight='balanced', y=ytrain)

    fold = clone(xgb)
    fold.fit(Xtrain, ytrain, model__sample_weight=weight)
    predicted = fold.predict(Xvalid)
    folds.append(f1_score(yvalid, predicted, average='macro'))

scores['XGBoost'] = np.array(folds)

for name, score in scores.items():
    print(f"{name}: {score.mean():.4f} +/- {score.std():.4f}")
```

### Fit and evaluate on the test set

```python
from sklearn.metrics import accuracy_score, balanced_accuracy_score, classification_report

lr.fit(X_train, y_train_encoded)
rf.fit(X_train, y_train_encoded)
xgb.fit(X_train, y_train_encoded, model__sample_weight=weights)

models = {
    'Logistic Regression': lr,
    'Random Forest': rf,
    'XGBoost': xgb
}

results = []
for name, model in models.items():
    predicted = model.predict(X_test)
    results.append({
        'Model': name,
        'Accuracy': accuracy_score(y_test_encoded, predicted),
        'F1-Macro': f1_score(y_test_encoded, predicted, average='macro'),
        'Balanced Accuracy': balanced_accuracy_score(y_test_encoded, predicted)
    })
    print(f"\n{name}")
    print(classification_report(y_test_encoded, predicted, target_names=encoder.classes_))

table = pd.DataFrame(results).sort_values('F1-Macro', ascending=False).reset_index(drop=True)
table
```

```python
from sklearn.metrics import confusion_matrix

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
for ax, (name, model) in zip(axes, models.items()):
    matrix = confusion_matrix(y_test_encoded, model.predict(X_test))
    sns.heatmap(matrix, annot=True, fmt='d', cmap='Blues',
                xticklabels=encoder.classes_, yticklabels=encoder.classes_, ax=ax)
    ax.set_title(name)
    ax.set_xlabel('Predicted'); ax.set_ylabel('Actual')
plt.tight_layout()
plt.show()
```

### Compare & pick the best model

```python
best = table.iloc[0]['Model']
pipeline = models[best]
print(f"Best model going into tuning: {best}")
table
```

### Hyperparameter tuning

```python
from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import loguniform, randint

spaces = {
    'Logistic Regression': {'model__C': loguniform(1e-3, 1e2)},
    'Random Forest': {
        'model__n_estimators': randint(200, 600),
        'model__max_depth': [None, 10, 20, 30],
        'model__min_samples_split': randint(2, 11),
        'model__min_samples_leaf': randint(1, 5),
        'model__max_features': ['sqrt', 'log2']
    },
    'XGBoost': {
        'model__n_estimators': randint(200, 600),
        'model__max_depth': randint(3, 9),
        'model__learning_rate': loguniform(1e-2, 3e-1),
        'model__subsample': [0.7, 0.8, 0.9, 1.0],
        'model__colsample_bytree': [0.7, 0.8, 0.9, 1.0]
    }
}

search = RandomizedSearchCV(pipeline, spaces[best],
                            n_iter=25, scoring='f1_macro', cv=kfold, random_state=42, n_jobs=-1)

if best == 'XGBoost':
    search.fit(X_train, y_train_encoded, model__sample_weight=weights)
else:
    search.fit(X_train, y_train_encoded)

print("Best CV f1_macro:", round(search.best_score_, 4))
print("Best params:", search.best_params_)
```

```python
tuned = search.best_estimator_
predictions = tuned.predict(X_test)

print(f"Final model: {best} (tuned)\n")
print(classification_report(y_test_encoded, predictions, target_names=encoder.classes_))
print("F1-Macro:", round(f1_score(y_test_encoded, predictions, average='macro'), 4))
print("Balanced Accuracy:", round(balanced_accuracy_score(y_test_encoded, predictions), 4))
```

### Feature importance

```python
names = tuned.named_steps['preprocessing'].named_steps['preprocess'].get_feature_names_out()
estimator = tuned.named_steps['model']

if hasattr(estimator, 'feature_importances_'):
    importances = pd.Series(estimator.feature_importances_, index=names)
else:
    importances = pd.Series(np.abs(estimator.coef_).mean(axis=0), index=names)
importances = importances.sort_values(ascending=False)

plt.figure(figsize=(8, 6))
sns.barplot(x=importances.values, y=importances.index, color='steelblue')
plt.title(f'Feature Importance - {best} (tuned)')
plt.tight_layout()
plt.show()
importances
```
